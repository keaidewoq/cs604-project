```python
REPO_URL = "https://github.com/keaidewoq/cs604-project.git"
REPO_DIR = "/content/repo"
DATA_DIR = "/content/data"
# Must match a folder (or shortcut) in My Drive — not the browser URL.
DRIVE_OUT = "/content/drive/MyDrive/CS604_RUNS"
BRANCH = "ana-branch"
METHOD = "adapter_w96_336_recipe"  # adapter_w96_336_recipe | adaptformer_w192_336_recipe | lora_r24_336_recipe
TEAM = "08"
PROTO_PT = f"{DRIVE_OUT}/prototypes_336.pt"  # must match the config's prototype_file
# lr pilot before the full run (needs a pilot: block in the config)
RUN_PILOT = True
PILOT_DIR = ""  # paste an earlier pilot_* folder on Drive to reuse its choice instead of rerunning
PILOT_DATA_DIR = "/content/pilot_data"  # local disk; label files only, images are symlinked
# End the runtime after the last cell, or as soon as training fails.
AUTO_DISCONNECT = True


def end_runtime(reason):
    """Flush Drive writes, then release the GPU so it stops using compute units."""
    print(reason, flush=True)
    if not AUTO_DISCONNECT:
        return
    from google.colab import drive, runtime
    try:
        drive.flush_and_unmount()  # without this the last checkpoint can be lost
        print("Drive flushed and unmounted", flush=True)
    except Exception as error:
        print(f"Drive flush failed: {error}", flush=True)
    runtime.unassign()
```

## 4c. lr pilot (optional)

Runs only if `RUN_PILOT = True`. Each lr in the config's `pilot.sweep` gets a short run on a 1,000-class subset (3 epochs), scored on a labelled 5-per-class holdout from those classes, in a separate process so memory is freed afterwards. Writes `pilot_sweep.csv`, `pilot_summary.json` and `best_config.yaml` to a `pilot_*` folder on Drive; the full run then trains with `best_config.yaml`.

Check the table before the full run starts: if the chosen lr is the highest one tried, a higher lr might do better (extend the sweep next time). The holdout accuracy is optimistic in absolute terms (the 336 prototypes were built from all training images, including these holdout images), but it is the same for every lr, so the ranking holds.

```python
import json
import os
import subprocess
import sys
import time
from pathlib import Path

import pandas as pd

TRAIN_CONFIG = f"configs/{METHOD}.yaml"
if RUN_PILOT:
    if PILOT_DIR:
        sweep_dir = Path(PILOT_DIR)
        print("Reusing pilot:", sweep_dir)
    else:
        pilot_started = time.time()
        proc = subprocess.Popen(
            [sys.executable, "-u", "scripts/pilot_sweep.py", "--config", TRAIN_CONFIG,
             "--data_root", DATA_DIR, "--output_dir", DRIVE_OUT, "--pilot_data_dir", PILOT_DATA_DIR],
            stdout=subprocess.PIPE, stderr=subprocess.STDOUT, text=True, bufsize=1,
            env={**os.environ, "PYTHONPATH": REPO_DIR},
        )
        for line in proc.stdout:
            print(line, end="", flush=True)
        if proc.wait() != 0:
            end_runtime(f"pilot failed (exit {proc.returncode}); see the output above")
            raise SystemExit("pilot failed")
        pilots = [p for p in Path(DRIVE_OUT).glob("pilot_*")
                  if p.stat().st_mtime >= pilot_started and (p / "pilot_summary.json").is_file()]
        assert pilots, "Pilot finished but no pilot_summary.json was found"
        sweep_dir = max(pilots, key=lambda p: p.stat().st_mtime)
    display(pd.read_csv(sweep_dir / "pilot_sweep.csv").drop(columns=["run_dir"], errors="ignore"))
    summary = json.loads((sweep_dir / "pilot_summary.json").read_text())
    print("Chosen:", summary["chosen"], "|", summary["reason"])
    TRAIN_CONFIG = str(sweep_dir / "best_config.yaml")
print("Full run will train with", TRAIN_CONFIG)
```

## 8. Inference latency (from `colab_launcher_qq.ipynb`)

Load the trained checkpoint selected in section 6. Measure forward-pass latency on synthetic images at batch sizes 1 and 64, excluding preprocessing and transfers. Use 30 warm-ups and 200 timed iterations. Save results to `inference_latency.json`. This runs outside the training timer; keep GPU and measurement settings consistent between experiments.


```python
import json
import torch

from src.utils import load_config, apply_overrides, set_seed
from src.model import build_model
from src.predict import load_checkpoint
from scripts.benchmark_latency import measure

assert torch.cuda.is_available(), "A GPU runtime is required."
cfg = load_config(RUN_DIR / "config.yaml")
# Partial PEFT checkpoints require the original pretrained backbone.
checkpoint_info = torch.load(CKPT, map_location="cpu", weights_only=False)
partial = bool(checkpoint_info.get("trainable_only", False))
del checkpoint_info
apply_overrides(cfg, pretrained=partial)
set_seed(int(cfg.seed))

device = torch.device("cuda")
model = build_model(cfg).to(device)
load_checkpoint(model, CKPT, device)
model.eval()

warmup = 30
iterations = 200
use_bf16 = bool(cfg.bf16)
results = {
    "checkpoint": str(CKPT),
    "gpu": torch.cuda.get_device_name(0),
    "gpu_capacity_GiB": torch.cuda.get_device_properties(0).total_memory / 1024**3,
    "torch_version": str(torch.__version__),
    "precision": "bf16" if use_bf16 else "fp32",
    "img_size": int(cfg.img_size),
    "warmup_iterations": warmup,
    "timed_iterations": iterations,
    "scope": "Forward pass on synthetic inputs; excludes loading, preprocessing and transfers",
    "measurements": [],
}

try:
    for batch_size in (1, 64):
        batch = torch.randn(
            batch_size, 3, int(cfg.img_size), int(cfg.img_size), device=device,
        )
        with torch.autocast(device_type="cuda", dtype=torch.bfloat16, enabled=use_bf16):
            ms_per_image, _ = measure(model, batch, device, warmup, iterations)
        results["measurements"].append({
            "batch_size": batch_size,
            "ms_per_image": ms_per_image,
            "ms_per_batch": ms_per_image * batch_size,
        })
        del batch
finally:
    del model
    torch.cuda.empty_cache()

output = RUN_DIR / "inference_latency.json"
output.write_text(json.dumps(results, indent=2), encoding="utf-8")
print(json.dumps(results, indent=2))
print("Saved:", output)
```
